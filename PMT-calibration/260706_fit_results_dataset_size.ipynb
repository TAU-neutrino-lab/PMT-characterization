```python
from pathlib import Path
import re

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import yaml

from scipy.optimize import curve_fit

from pmt.fitresults import *

plt.rcParams.update({
    "figure.figsize": (8, 5),
    "axes.grid": True,
    "grid.alpha": 0.60,
})
```

```python
# load files
fit_res_dir = Path("plots/260706/fit/fit_results")

files = sorted(fit_res_dir.glob("*led100_poisson.yaml"))
df_poisson = pd.DataFrame([load_fit_yaml(f) for f in files])

files = sorted(fit_res_dir.glob("*led100_bellamy.yaml"))
df_bellamy = pd.DataFrame([load_fit_yaml(f) for f in files])

files = sorted(fit_res_dir.glob("*_150k_poisson.yaml"))
df_poisson150 = pd.DataFrame([load_fit_yaml(f) for f in files])

files = sorted(fit_res_dir.glob("*_150k_bellamy.yaml"))
df_bellamy150 = pd.DataFrame([load_fit_yaml(f) for f in files])
```

# Poisson

```python
fig, axes = plt.subplots(2, 3, figsize=(20, 10))
axes = axes.ravel()

xtitle = "Input voltage  [V]"
formt = 'o-'

bounds = True

parms = ["q1_mV_ns", "sigma1_mV_ns", "n_total", "q0_mV_ns", "sigma0_mV_ns", "chi2_ndof"]
ytitles = [r"$Q_1$", r"$\sigma_1$", r"N events", r"$Q_0$", r"$\sigma_0$", r"$\chi^2 / ndof$"]

i = 0

for par, ytitle in zip(parms, ytitles):
    title = ytitle
    plot_parameter( df_poisson,    par, ax=axes[i], x="HV", xlabel=xtitle, ylabel=ytitle, title=title, label="Poisson 300k", show_bounds=bounds, fmt=formt, markersize=3)
    plot_parameter( df_poisson150, par, ax=axes[i], x="HV", xlabel=xtitle, ylabel=ytitle, title=title, label="Poisson 150k", show_bounds=bounds, fmt=formt, markersize=3)
    i+=1

```
