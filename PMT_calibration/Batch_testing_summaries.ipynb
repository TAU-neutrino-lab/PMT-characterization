# Batch testing summaries

Run the full diagnostic plot set from `testing.ipynb` once per voltage cache and save one multipage PDF per acquisition. Figures are closed after each PDF page so voltages are processed sequentially without retaining waveform plots in RAM.

```python
from pathlib import Path
import importlib

import pmt.testing_summary as testing_summary
testing_summary = importlib.reload(testing_summary)

# Add date tags to process several dated cache directories in one run.
date_tags = ["16_8"]  # for example: ["3_8", "16_8"]
testing_fit_data_root = Path("plots/dark_counts/fit_data")
testing_summary_output_root = Path("plots")
testing_cache_glob = "*_Dark_20microV_trig_df.pkl"
testing_notebook_path = Path("testing.ipynb")
```

```python
testing_summary_records = {}
for date_tag in date_tags:
    testing_summary_records[date_tag] = testing_summary.generate_testing_summaries(
        testing_fit_data_root / date_tag,
        testing_summary_output_root / f"dark_counts_{date_tag}",
        cache_glob=testing_cache_glob,
        testing_notebook_path=testing_notebook_path,
    )
testing_summary_records
```
